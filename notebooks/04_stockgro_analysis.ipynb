# 3 — PREDICTED PRICES TABLE TEMPLATE

```python
prediction_table = pd.DataFrame({
    "Stock": [
        "Reliance",
        "Infosys",
        "HDFC Bank",
        "Tata Steel",
        "Sun Pharma"
    ],

     "Predicted_Day1": [1350, 1125, 745, 214, 1835],
    "Actual_Day1": [1349.14, 1126.62, 739.79, 209.55, 1834.73],
    "Predicted_Day2": [1360, 1130, 748, 216, 1825],
    "Actual_Day2": [1338.11, 1115.42, 743.37, 216.77, 1810.54]
})

prediction_table
```

# 4 — ERROR CALCULATION TEMPLATE

```python
prediction_table['Day1_Error_%'] = (
    abs(
        prediction_table['Predicted_Day1']
        -
        prediction_table['Actual_Day1']
    )
    /
    prediction_table['Actual_Day1']
) * 100

prediction_table['Day2_Error_%'] = (
    abs(
        prediction_table['Predicted_Day2']
        -
        prediction_table['Actual_Day2']
    )
    /
    prediction_table['Actual_Day2']
) * 100

prediction_table
```

# 5 — PORTFOLIO RETURN TEMPLATE

```python
initial_capital = 998395.83

final_portfolio_value = 987409.88

portfolio_return = (
    (
        final_portfolio_value
        -
        initial_capital
    )
    /
    initial_capital
) * 100

print(f"Portfolio Return: {portfolio_return:.2f}%")
```
